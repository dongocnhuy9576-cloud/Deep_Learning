```python
# Bài 6: Phép toán và nhân ma trận

a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:")
print(a)

print("\nb:")
print(b)

print("\na + b:")
print(a + b)

print("\na / b:")
print(a / b)

print("\na^2:")
print(a ** 2)

print("\na * b:")
print(a * b)

print("\na @ b.T:")
print(a @ b.T)

print("\nShape của a * b:")
print((a * b).shape)

print("\nShape của a @ b.T:")
print((a @ b.T).shape)

print("\nNhận xét:")
print("a * b là phép nhân từng phần tử tương ứng.")
print("a @ b.T là phép nhân ma trận.")
```

```python
# Bài 7: Các hàm thống kê trên Tensor

a = torch.randn(3, 4)

print("Tensor a:")
print(a)

print("\nMean theo cột - dim=0:")
print(torch.mean(a, dim=0))

print("\nStandard deviation theo cột - dim=0:")
print(torch.std(a, dim=0))

print("\nCumulative sum theo cột - dim=0:")
print(torch.cumsum(a, dim=0))

print("\nMean theo hàng - dim=1:")
print(torch.mean(a, dim=1))

print("\nStandard deviation theo hàng - dim=1:")
print(torch.std(a, dim=1))

print("\nCumulative sum theo hàng - dim=1:")
print(torch.cumsum(a, dim=1))

print("\nNhận xét:")
print("dim=0: tính toán theo chiều dọc, kết quả theo từng cột.")
print("dim=1: tính toán theo chiều ngang, kết quả theo từng hàng.")
```

```python
# Bài 8: Autograd cơ bản

# Hàm y = x^2
x = torch.tensor(3.6, requires_grad=True)

y = x * x

y.backward()

print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

print("\nTính thủ công:")
print("y = x^2")
print("dy/dx = 2x")
print("Tại x = 3.6:")
print("dy/dx =", 2 * 3.6)


# Hàm y = 3*x^2 + 2*x + 1
x = torch.tensor(3.6, requires_grad=True)

y = 3 * x**2 + 2 * x + 1

y.backward()

print("\n------------------------------")
print("Hàm y = 3*x^2 + 2*x + 1")

print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

print("\nTính thủ công:")
print("dy/dx = 6x + 2")
print("Tại x = 3.6:")
print("dy/dx =", 6 * 3.6 + 2)
```

```python
# Bài 9: Gradient Descent tìm hệ số đa thức

# Đa thức cần tìm:
# y = x^2 + 2x + 3

polynomial = np.poly1d([1, 2, 3])

N = 20

# Sinh dữ liệu
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

# Tạo ma trận [x^2, x, 1]
XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

# Khởi tạo các hệ số ngẫu nhiên
w = torch.randn(3, 1, requires_grad=True)

# Chuyển sang Tensor
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

# Optimizer
optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

# Huấn luyện
for epoch in range(1000):

    # Xóa gradient cũ
    optimizer.zero_grad()

    # Dự đoán
    y_pred = x @ w

    # Tính MSE
    mse = torch.mean(torch.square(y - y_pred))

    # Tính gradient
    mse.backward()

    # Cập nhật hệ số
    optimizer.step()

print("\nLearned coefficients:")
print(w)

print("\nHệ số thực tế:")
print("[1, 2, 3]")
```

```python
# Bài 10: Giải hệ phương trình bằng Autograd

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000

optimizer = torch.optim.NAdam(
    [A, B, C, D],
    lr=0.01
)

for epoch in range(EPOCHS):

    # Bốn phương trình
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    # Tổng sai số bình phương
    sqerr = (
        y1*y1 +
        y2*y2 +
        y3*y3 +
        y4*y4
    )

    # Xóa gradient
    optimizer.zero_grad()

    # Tính gradient
    sqerr.backward()

    # Cập nhật A, B, C, D
    optimizer.step()

print("Nghiệm tìm được:")

print("A =", A.item())
print("B =", B.item())
print("C =", C.item())
print("D =", D.item())

print("\nKiểm tra hệ phương trình:")

print("A + B =", (A + B).item())
print("C - D =", (C - D).item())
print("A + C =", (A + C).item())
print("B - D =", (B - D).item())

print("\nNghiệm chính xác:")
print("A = 5")
print("B = 4")
print("C = 3")
print("D = 2")
```

```python

# Đa thức y = 2x^2 - 3x + 5
polynomial = np.poly1d([2, -3, 5])

# Tạo dữ liệu
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

# Tạo ma trận [x^2, x, 1]
XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

# Chuyển dữ liệu sang Tensor
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

# Khởi tạo 3 hệ số ngẫu nhiên
w = torch.randn(3, 1, requires_grad=True)

# Tạo Optimizer
optimizer = torch.optim.NAdam([w], lr=0.01)

print("Hệ số ban đầu:")
print(w.detach())

# Huấn luyện
for epoch in range(1000):
    optimizer.zero_grad()

    y_pred = x @ w

    mse = torch.mean(torch.square(y - y_pred))

    mse.backward()

    optimizer.step()

print("\nHệ số sau khi huấn luyện:")
print(w.detach())

print("\nHệ số thực tế:")
print("[2, -3, 5]")

print("\nMSE:")
print(mse.item())
```
